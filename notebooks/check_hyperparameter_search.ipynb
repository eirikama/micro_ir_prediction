```python
import optuna

domain = 'pcuk'
storage_path = f"sqlite:///../sweeps/{domain}_sweep.db"
study = optuna.load_study(study_name=f"{domain}_augmentation_sweep", storage=storage_path)

print(f"Number of trials: {len(study.trials)}")
print(f"Best value: {study.best_value}")
print(f"Best params:")
for k, v in study.best_params.items():
    print(f"  {k}: {v}")
print(f"Best trial number: {study.best_trial.number}")
```
